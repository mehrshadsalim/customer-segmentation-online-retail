```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="pastel", font_scale=1.1)

# Load cleaned transaction data (Phase 1 output)
tx = pd.read_csv("../data/online_retail_clean.csv", encoding="ISO-8859-1")

# Standardise columns (just in case)
tx.columns = [c.strip().lower().replace(" ", "") for c in tx.columns]

tx["date"] = pd.to_datetime(tx["date"], errors="coerce")

# Load RFM with final clusters (Phase 2 output)
rfm = pd.read_csv("../reports/rfm_clusters.csv")

print("Transactions:", tx.shape)
print("RFM with clusters:", rfm.shape)

tx.head(), rfm.head()
```

```python
tx_cluster = tx.merge(
    rfm[["customerno", "recency", "frequency", "monetary", "cluster"]],
    on="customerno",
    how="inner"
)

print("Transactions with clusters:", tx_cluster.shape)
tx_cluster.head()
```

```python
#Step 2 – Trend Analysis by Cluster
```

```python
tx_cluster["month"] = tx_cluster["date"].dt.to_period("M").dt.to_timestamp()

tx_cluster[["date","month","cluster"]].head()
```

```python
#Monthly revenue per cluster
monthly_revenue_cluster = (
    tx_cluster
    .groupby(["month","cluster"])["totalprice"]
    .sum()
    .reset_index()
    .rename(columns={"totalprice":"monthly_revenue"})
)

monthly_revenue_cluster.head()
```

```python
#نمودار: درآمد ماهانه به تفکیک خوشه
plt.figure(figsize=(10, 6))
for c in sorted(monthly_revenue_cluster["cluster"].unique()):
    subset = monthly_revenue_cluster[monthly_revenue_cluster["cluster"] == c]
    plt.plot(subset["month"], subset["monthly_revenue"], marker="o", label=f"Cluster {c}")

plt.title("Monthly Revenue per Cluster")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Active customers per cluster
monthly_active_customers = (
    tx_cluster
    .groupby(["month","cluster"])["customerno"]
    .nunique()
    .reset_index()
    .rename(columns={"customerno":"active_customers"})
)

monthly_active_customers.head()
```

```python
plt.figure(figsize=(10, 6))
for c in sorted(monthly_active_customers["cluster"].unique()):
    subset = monthly_active_customers[monthly_active_customers["cluster"] == c]
    plt.plot(subset["month"], subset["active_customers"], marker="o", label=f"Cluster {c}")

plt.title("Monthly Active Customers per Cluster")
plt.xlabel("Month")
plt.ylabel("Number of active customers")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Country / Region Analysis per Cluster
country_total = (
    tx_cluster
    .groupby("country")["totalprice"]
    .sum()
    .rename("country_revenue")
    .reset_index()
)

country_total.sort_values("country_revenue", ascending=False).head()
```

```python
cluster_country_rev = (
    tx_cluster
    .groupby(["country","cluster"])["totalprice"]
    .sum()
    .reset_index()
)

cluster_country_rev = cluster_country_rev.merge(country_total, on="country", how="left")
cluster_country_rev["share_in_country"] = 100 * cluster_country_rev["totalprice"] / cluster_country_rev["country_revenue"]

cluster_country_rev.head()
```

```python
top_countries = (
    country_total
    .sort_values("country_revenue", ascending=False)
    .head(10)["country"]
)

cc_top = cluster_country_rev[cluster_country_rev["country"].isin(top_countries)]
```

```python
pivot_comp = cc_top.pivot(
    index="cluster",
    columns="country",
    values="share_in_country"
).fillna(0)

plt.figure(figsize=(12, 5))
sns.heatmap(pivot_comp, cmap="coolwarm", annot=True, fmt=".1f")
plt.title("Cluster Composition per Country (% of Country Revenue)")
plt.xlabel("Country")
plt.ylabel("Cluster")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Classical RFM Score vs K-Means Cluster
rfm_scores = rfm.copy()

rfm_scores["recency_score"] = pd.qcut(
    rfm_scores["recency"], 5, labels=[5,4,3,2,1]
).astype(int)

rfm_scores["frequency_score"] = pd.qcut(
    rfm_scores["frequency"].rank(method="first"), 5, labels=[1,2,3,4,5]
).astype(int)

rfm_scores["monetary_score"] = pd.qcut(
    rfm_scores["monetary"].rank(method="first"), 5, labels=[1,2,3,4,5]
).astype(int)

rfm_scores["rfm_score"] = (
    rfm_scores["recency_score"] +
    rfm_scores["frequency_score"] +
    rfm_scores["monetary_score"]
)

rfm_scores[["customerno","cluster","recency_score","frequency_score","monetary_score","rfm_score"]].head()
```

```python
rfm_cluster_ct = pd.crosstab(rfm_scores["cluster"], rfm_scores["rfm_score"])
rfm_cluster_ct
```

```python
plt.figure(figsize=(10, 5))
sns.heatmap(rfm_cluster_ct, annot=True, fmt="d", cmap="viridis")
plt.title("RFM Score vs K-Means Cluster")
plt.xlabel("RFM Score")
plt.ylabel("Cluster")
plt.tight_layout()
plt.show()
```

```python
# ==============================
# Step 5 – CLV-style Ranking per Segment (CLEAN VERSION)
# ==============================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="pastel", font_scale=1.1)

# 1) از RFM با خوشه‌ها دوباره KPIها را می‌سازیم
# اگر قبلاً رِفِرِنس rfm داری، از همون استفاده کن؛
# اگر نه، دوباره لودش کن:
# rfm = pd.read_csv("../reports/rfm_clusters.csv")

cluster_kpis = (
    rfm.groupby("cluster")
       .agg(
           avg_recency   = ("recency", "mean"),
           avg_frequency = ("frequency", "mean"),
           avg_monetary  = ("monetary", "mean")
       )
       .reset_index()
)

print("Clean cluster_kpis:")
print(cluster_kpis)

# 2) محاسبه‌ی CLV-style proxy
cluster_kpis["clv_proxy"] = (
    cluster_kpis["avg_frequency"] * cluster_kpis["avg_monetary"]
) / (cluster_kpis["avg_recency"] + 1e-9)

cluster_kpis_sorted = cluster_kpis.sort_values("clv_proxy", ascending=False)
print("\nClusters ranked by CLV proxy:")
print(cluster_kpis_sorted[["cluster","avg_recency","avg_frequency","avg_monetary","clv_proxy"]])

# 3) نمودار
plt.figure(figsize=(6, 4))
sns.barplot(
    data=cluster_kpis_sorted,
    x="cluster",
    y="clv_proxy"
)
plt.title("CLV-style proxy per cluster")
plt.xlabel("Cluster")
plt.ylabel("CLV proxy")
plt.tight_layout()
plt.show()
```

```python

sns.set(style="whitegrid", palette="pastel", font_scale=1.1)

# تعداد مشتری‌ها در هر خوشه (از خودت گرفتیم)
cluster_counts = rfm["cluster"].value_counts().rename("customers")
print("Customer count per cluster:")
display(cluster_counts)

# استفاده از cluster_kpis از Step 5
kpis = cluster_kpis.set_index("cluster")[["avg_monetary", "avg_recency", "avg_frequency", "clv_proxy"]]

# ترکیب KPIها + تعداد مشتری‌ها
cluster_data = pd.concat([kpis, cluster_counts], axis=1)

print("Cluster Data for Simulation:")
display(cluster_data)
```

```python
# سناریوهای مارکتینگ با نرخ تبدیل مختلف
scenarios = {
    "weak": 0.01,       # 1% conversion
    "base": 0.05,       # 5%
    "optimistic": 0.10  # 10%
}

print("Scenarios (conversion rates):", scenarios)
```

```python
results = []

for scenario_name, rate in scenarios.items():
    for cluster in cluster_data.index:
        avg_monetary = cluster_data.loc[cluster, "avg_monetary"]
        n_customers  = cluster_data.loc[cluster, "customers"]

        projected_revenue = n_customers * rate * avg_monetary

        results.append({
            "scenario": scenario_name,
            "cluster": cluster,
            "conversion_rate": rate,
            "projected_revenue": projected_revenue
        })

results_df = pd.DataFrame(results)

print("Simulation results:")
display(results_df)
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=results_df,
    x="cluster",
    y="projected_revenue",
    hue="scenario"
)
plt.title("Projected Revenue per Cluster Across Scenarios")
plt.xlabel("Cluster")
plt.ylabel("Projected Revenue (£)")
plt.tight_layout()
plt.show()
```
